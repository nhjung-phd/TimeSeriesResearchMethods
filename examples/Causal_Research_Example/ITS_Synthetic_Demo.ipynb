# Chapter 5 · Synthetic ITS Demonstration

**연구질문:** 알려진 개입 이후 즉각 수준·추세 변화가 추정되는가?

교육용 합성자료로 인과 추정기 코드를 검증합니다. **실제 사건 효과가 아닙니다.**

$$Y_t=\beta_0+\beta_1 t+\beta_2 Post_t+\beta_3 TimeAfter_t+\epsilon_t$$

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import statsmodels.api as sm
rng = np.random.default_rng(2026)
n=120; t=np.arange(n); t0=60
post=(t>=t0).astype(int)
after=np.maximum(t-t0,0)
y = 100 + .2*t - 4*post + .12*after + rng.normal(0, 1.5, n)
X=sm.add_constant(pd.DataFrame({'trend':t,'post':post,'time_after':after}))
fit=sm.OLS(y,X).fit(cov_type='HAC',cov_kwds={'maxlags':6})
print(fit.summary())
print('True synthetic level shift -4; true slope change +0.12')
```

```python
cf_X=X.copy();cf_X[['post','time_after']]=0
counterfactual=fit.predict(cf_X)
plt.figure(figsize=(11,4))
plt.plot(t,y,label='Observed synthetic series',alpha=.75)
plt.plot(t,counterfactual,label='Estimated no-intervention counterfactual',linestyle='--')
plt.axvline(t0,color='black',linestyle=':',label='Intervention')
plt.title('Synthetic ITS: observed vs counterfactual')
plt.xlabel('Time index');plt.ylabel('Synthetic outcome')
plt.legend();plt.show()
```

## 해석·검증
- level shift와 slope change의 추정치·표준오차·신뢰구간 비교
- 동일 시점에 추가 사건이 있었다면 ITS만으로 인과성을 주장하기 어려움
- placebo 개입시점, 사전기간 변경, HAC lag 변경이 심화실습
